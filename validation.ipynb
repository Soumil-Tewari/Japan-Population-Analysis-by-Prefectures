```python
print("Negative: ",(df[pop]<0).sum())
print("Zero: ",(df[pop]==0).sum())
```

```python
age_cols = [
    "Percentage distribution of total population (0-14)【%】",
    "Percentage distribution of total population (15-64)【%】",
    "Percentage distribution of total population (65 and over)【%】"
]
df["age_share_sum"]=df[age_cols].sum(axis=1,min_count=3)
print(df[age_cols].isna().all(axis=1).sum())

print(df["age_share_sum"].describe())
bad_age=df[df["age_share_sum"].notna() & ((df["age_share_sum"]<98.6)|(df["age_share_sum"]>101))]
print(df.loc[df["age_share_sum"].notna(),["Time","Region","age_share_sum"]].sort_values(by='age_share_sum').head(5))
print(df.loc[df["age_share_sum"].notna(),"age_share_sum"].sort_values().tail(5))
print("Age-share anomalies",len(bad_age))

for col in age_cols:
    print(col,
          "negative: ",(df[col]<=0).sum(),
          "above 100: ",(df[col]>=100).sum()
          )

```

```python
df=df.sort_values(["Region Code","Time"])
df["Population change"]=(df.groupby("Region Code")[pop].diff())
df["Net Migration"] = (
    df["Total in-migrants (overall)【person】"]
    - df["Total out-migrants (overall)【person】"]
)
print(df["Net Migration"].describe())
df["Expected Population Change"]=(
   df["Live births【person】"] - df["Deaths【person】"]  + df["Net Migration"]
)
accounting=df.dropna(
    subset=[
        "Population change",
        "Live births【person】",
        "Deaths【person】",
        "Net Migration"
    ],how="any").copy()
accounting["Accounting Difference"]=(accounting["Population change"]-accounting["Expected Population Change"])
print("Population change NA:", df["Population change"].isna().sum())
print("Accounting rows:", len(accounting))

```
